```python
import pandas as pd

# 1. Convert to string (important)
df[' Income '] = df[' Income '].astype(str)

# 2. Remove $, commas, and spaces
df[' Income '] = df[' Income '].str.replace(r'[$,]', '', regex=True).str.strip()

# 3. Replace known bad values with NaN
df[' Income '] = df[' Income '].replace(['nan', 'None', '', '?'], pd.NA)

# 4. Convert to numeric
df[' Income '] = pd.to_numeric(df[' Income '], errors='coerce')
```

```python
df = df.dropna(subset=[' Income ']).reset_index(drop=True)
```
